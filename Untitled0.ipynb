```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist

# =========================================================
# 1. GRU & MGU CELL IMPLEMENTATIONS
# =========================================================

def init_gru_params(input_dim, hidden_dim):
    glorot = tf.initializers.GlorotUniform()
    return {
        "Wz": glorot([input_dim, hidden_dim]),
        "Uz": glorot([hidden_dim, hidden_dim]),
        "bz": tf.zeros([hidden_dim]),

        "Wr": glorot([input_dim, hidden_dim]),
        "Ur": glorot([hidden_dim, hidden_dim]),
        "br": tf.zeros([hidden_dim]),

        "Ws": glorot([input_dim, hidden_dim]),
        "Us": glorot([hidden_dim, hidden_dim]),
        "bs": tf.zeros([hidden_dim]),
    }

def init_mgu_params(input_dim, hidden_dim):
    glorot = tf.initializers.GlorotUniform()
    return {
        "Wf": glorot([input_dim, hidden_dim]),
        "Uf": glorot([hidden_dim, hidden_dim]),
        "bf": tf.zeros([hidden_dim]),

        "Ws": glorot([input_dim, hidden_dim]),
        "Us": glorot([hidden_dim, hidden_dim]),
        "bs": tf.zeros([hidden_dim]),
    }

# Small functions can be tf.function
@tf.function
def gru_step(x_t, s_prev, p):
    z = tf.sigmoid(x_t @ p["Wz"] + s_prev @ p["Uz"] + p["bz"])
    r = tf.sigmoid(x_t @ p["Wr"] + s_prev @ p["Ur"] + p["br"])
    s_hat = tf.tanh(x_t @ p["Ws"] + (r * s_prev) @ p["Us"] + p["bs"])
    return (1 - z) * s_prev + z * s_hat

@tf.function
def mgu_step(x_t, s_prev, p):
    f = tf.sigmoid(x_t @ p["Wf"] + s_prev @ p["Uf"] + p["bf"])
    s_hat = tf.tanh(x_t @ p["Ws"] + (f * s_prev) @ p["Us"] + p["bs"])
    return (1 - f) * s_prev + f * s_hat

# =========================================================
# 2. RNN UNROLLING (eager-friendly)
# =========================================================
def run_rnn(x, params, step_fn):
    batch = tf.shape(x)[0]
    time_steps = tf.shape(x)[1]
    hidden_dim = params[list(params.keys())[0]].shape[-1]

    s = tf.zeros([batch, hidden_dim], dtype=tf.float32)
    for t in range(time_steps):
        s = step_fn(x[:, t, :], s, params)
    return s

# =========================================================
# 3. RNN CLASSIFIER
# =========================================================
class RNNClassifier(tf.keras.Model):
    def __init__(self, input_dim, hidden_dim, output_dim, cell_type):
        super().__init__()
        self.cell_type = cell_type
        init_params = init_gru_params(input_dim, hidden_dim) if cell_type=="GRU" \
                       else init_mgu_params(input_dim, hidden_dim)

        self.params = {}
        for name, val in init_params.items():
            self.params[name] = self.add_weight(
                name=name,
                shape=val.shape,
                initializer=tf.initializers.GlorotUniform()
                if not name.startswith("b") else tf.zeros_initializer(),
                trainable=True
            )

        self.Wo = self.add_weight(name="Wo", shape=[hidden_dim, output_dim],
                                  initializer=tf.initializers.GlorotUniform())
        self.bo = self.add_weight(name="bo", shape=[output_dim],
                                  initializer=tf.zeros_initializer())

        self.step_fn = gru_step if cell_type == "GRU" else mgu_step

    def call(self, x):
        # Eager execution (no tf.function decorator here)
        h = run_rnn(x, self.params, self.step_fn)
        return h @ self.Wo + self.bo

# =========================================================
# 4. LOAD MNIST
# =========================================================
def load_mnist_sequence():
    (x_train, y_train), (x_test, y_test) = mnist.load_data()
    x_train = x_train.astype(np.float32) / 255.0
    x_test = x_test.astype(np.float32) / 255.0
    return x_train.reshape([-1,28,28]), y_train, x_test.reshape([-1,28,28]), y_test

# =========================================================
# 5. TRAINING LOOP
# =========================================================
def train(model, x_train, y_train, x_test, y_test, epochs=5, batch_size=256):
    optimizer = tf.keras.optimizers.Adam(1e-3)
    loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)

    train_losses, test_accs = [], []

    N = x_train.shape[0]
    for epoch in range(epochs):
        idx = np.random.permutation(N)
        x_train, y_train = x_train[idx], y_train[idx]

        epoch_loss = 0
        for i in range(0, N, batch_size):
            xb = tf.convert_to_tensor(x_train[i:i+batch_size])
            yb = tf.convert_to_tensor(y_train[i:i+batch_size])

            with tf.GradientTape() as tape:
                logits = model(xb)
                loss = loss_fn(yb, logits)

            grads = tape.gradient(loss, model.trainable_variables)
            optimizer.apply_gradients(zip(grads, model.trainable_variables))
            epoch_loss += float(loss)

        # Test accuracy
        logits_test = model(tf.convert_to_tensor(x_test))
        preds = tf.argmax(logits_test, axis=1).numpy()
        acc = np.mean(preds == y_test)

        train_losses.append(epoch_loss / (N // batch_size))
        test_accs.append(acc)
        print(f"Epoch {epoch+1} - Loss: {train_losses[-1]:.4f} | Acc: {acc:.4f}")

    return train_losses, test_accs

# =========================================================
# 6. RUN EXPERIMENT
# =========================================================
if __name__ == "__main__":
    x_train, y_train, x_test, y_test = load_mnist_sequence()

    print("\nTraining GRU Model...")
    gru_model = RNNClassifier(28, 64, 10, "GRU")
    gru_loss, gru_acc = train(gru_model, x_train, y_train, x_test, y_test)

    print("\nTraining MGU Model...")
    mgu_model = RNNClassifier(28, 64, 10, "MGU")
    mgu_loss, mgu_acc = train(mgu_model, x_train, y_train, x_test, y_test)

    # Plots
    plt.figure(figsize=(14,6))
    plt.subplot(1,2,1)
    plt.plot(gru_loss, label="GRU Loss")
    plt.plot(mgu_loss, label="MGU Loss")
    plt.title("Training Loss Comparison")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(gru_acc, label="GRU Accuracy")
    plt.plot(mgu_acc, label="MGU Accuracy")
    plt.title("Test Accuracy Comparison")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()
```
