```python
import time
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# -----------------------
# Keep original NumPy functions (must not delete)
# -----------------------
def batch_norm(x, gamma, beta, eps=1e-5):
    mean = np.mean(x, axis=0, keepdims=True)
    var = np.mean((x - mean) ** 2, axis=0, keepdims=True)
    x_hat = (x - mean) / np.sqrt(var + eps)
    return gamma * x_hat + beta

def layer_norm(x, gamma, beta, eps=1e-5):
    mean = np.mean(x, axis=1, keepdims=True)
    var = np.mean((x - mean) ** 2, axis=1, keepdims=True)
    x_hat = (x - mean) / np.sqrt(var + eps)
    return gamma * x_hat + beta

def weight_norm(v, g, eps=1e-8):
    v_norm = np.linalg.norm(v) + eps
    return g * (v / v_norm)

# -----------------------
# Dataset: Fashion-MNIST (matches screenshot)
# -----------------------
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
x_train = x_train.astype(np.float32) / 255.0
x_test = x_test.astype(np.float32) / 255.0
x_train = np.expand_dims(x_train, -1)  # (N,H,W,1)
x_test = np.expand_dims(x_test, -1)
NUM_CLASSES = 10
INPUT_SHAPE = x_train.shape[1:]

# Optionally reduce dataset for speed; set SUBSET=None to use all training data
SUBSET = 10000
if SUBSET:
    x_train = x_train[:SUBSET]
    y_train = y_train[:SUBSET]
    x_test = x_test[:2000]
    y_test = y_test[:2000]

print("Dataset: Fashion-MNIST. Input shape:", INPUT_SHAPE, "Train size:", x_train.shape[0])

# -----------------------
# Custom TensorFlow normalization implementations (basic ops only)
# -----------------------

# Wrapped tf_batch_norm into a Keras Layer
class CustomBatchNorm(tf.keras.layers.Layer):
    def __init__(self, name=None, eps=1e-5, **kwargs):
        super().__init__(name=name, **kwargs)
        self.eps = eps
        self.gamma = None
        self.beta = None

    def build(self, input_shape):
        num_channels = input_shape[-1]
        self.gamma = self.add_weight(
            name='gamma',
            shape=[num_channels],
            initializer=tf.initializers.ones(),
            trainable=True)
        self.beta = self.add_weight(
            name='beta',
            shape=[num_channels],
            initializer=tf.initializers.zeros(),
            trainable=True)
        super().build(input_shape)

    def call(self, x):
        axes = [0, 1, 2]  # batch, height, width
        mean = tf.reduce_mean(x, axis=axes, keepdims=True)
        var = tf.reduce_mean(tf.square(x - mean), axis=axes, keepdims=True)
        x_hat = (x - mean) / tf.sqrt(var + self.eps)
        gamma_r = tf.reshape(self.gamma, [1, 1, 1, -1])
        beta_r = tf.reshape(self.beta, [1, 1, 1, -1])
        return gamma_r * x_hat + beta_r

# Wrapped tf_layer_norm into a Keras Layer
class CustomLayerNorm(tf.keras.layers.Layer):
    def __init__(self, name=None, eps=1e-5, **kwargs):
        super().__init__(name=name, **kwargs)
        self.eps = eps
        self.gamma = None
        self.beta = None

    def build(self, input_shape):
        num_channels = input_shape[-1]
        self.gamma = self.add_weight(
            name='gamma',
            shape=[num_channels],
            initializer=tf.initializers.ones(),
            trainable=True)
        self.beta = self.add_weight(
            name='beta',
            shape=[num_channels],
            initializer=tf.initializers.zeros(),
            trainable=True)
        super().build(input_shape)

    def call(self, x):
        axes = [1, 2, 3]  # across features for each sample
        mean = tf.reduce_mean(x, axis=axes, keepdims=True)
        var = tf.reduce_mean(tf.square(x - mean), axis=axes, keepdims=True)
        x_hat = (x - mean) / tf.sqrt(var + self.eps)
        gamma_r = tf.reshape(self.gamma, [1, 1, 1, -1])
        beta_r = tf.reshape(self.beta, [1, 1, 1, -1])
        return gamma_r * x_hat + beta_r


class WeightNormDense(tf.keras.layers.Layer):
    """
    Dense with Weight Normalization reparameterization: w = (g/||v||) * v
    trains v and g directly. Uses pure TF ops to produce outputs.
    """
    def __init__(self, units, use_bias=True):
        super().__init__()
        self.units = int(units)
        self.use_bias = use_bias

    def build(self, input_shape):
        in_dim = int(input_shape[-1])
        initializer = tf.keras.initializers.GlorotUniform()
        v_init = initializer(shape=(in_dim, self.units))
        self.v = tf.Variable(initial_value=v_init, trainable=True, dtype=tf.float32, name='wn_v')
        g_init = tf.norm(v_init, axis=0)
        self.g = tf.Variable(initial_value=g_init, trainable=True, dtype=tf.float32, name='wn_g')
        if self.use_bias:
            self.b = tf.Variable(tf.zeros([self.units]), trainable=True, dtype=tf.float32)
        else:
            self.b = None

    def call(self, inputs):
        v_norm = tf.sqrt(tf.reduce_sum(tf.square(self.v), axis=0) + 1e-8)  # [units]
        scale = tf.reshape(self.g / v_norm, [1, -1])  # [1, units]
        w = self.v * scale  # [in_dim, units]
        out = tf.matmul(inputs, w)
        if self.b is not None:
            out = out + self.b
        return out

# -----------------------
# Model builders
# -----------------------
def build_baseline():
    inputs = tf.keras.Input(shape=INPUT_SHAPE)
    x = tf.keras.layers.Conv2D(32, 3, padding='same', activation=None)(inputs)
    x = tf.keras.layers.ReLU()(x)
    x = tf.keras.layers.MaxPool2D()(x)
    x = tf.keras.layers.Conv2D(64, 3, padding='same', activation=None)(x)
    x = tf.keras.layers.ReLU()(x)
    x = tf.keras.layers.MaxPool2D()(x)
    x = tf.keras.layers.Flatten()(x)
    x = tf.keras.layers.Dense(128, activation='relu')(x)
    logits = tf.keras.layers.Dense(NUM_CLASSES)(x)
    return tf.keras.Model(inputs, logits)

def build_custom(norm_type='batch'):
    """
    norm_type in {'batch','layer','weight'}
    For 'weight' we use WeightNormDense for the intermediate Dense layer.
    For 'batch' and 'layer' we apply the custom TF batch/layer norm after conv layers.
    """
    inputs = tf.keras.Input(shape=INPUT_SHAPE)
    x = inputs

    # Conv1
    x = tf.keras.layers.Conv2D(32, 3, padding='same', activation=None, use_bias=False)(x)
    if norm_type == 'batch':
        x = CustomBatchNorm(name='bn1')(x)
    elif norm_type == 'layer':
        x = CustomLayerNorm(name='ln1')(x)
    x = tf.keras.layers.ReLU()(x)
    x = tf.keras.layers.MaxPool2D()(x)

    # Conv2
    x = tf.keras.layers.Conv2D(64, 3, padding='same', activation=None, use_bias=False)(x)
    if norm_type == 'batch':
        x = CustomBatchNorm(name='bn2')(x)
    elif norm_type == 'layer':
        x = CustomLayerNorm(name='ln2')(x)
    x = tf.keras.layers.ReLU()(x)
    x = tf.keras.layers.MaxPool2D()(x)

    x = tf.keras.layers.Flatten()(x)
    if norm_type == 'weight':
        # WeightNormDense then ReLU
        x = WeightNormDense(128)(x)
        x = tf.keras.layers.ReLU()(x)
        logits = tf.keras.layers.Dense(NUM_CLASSES)(x)
    else:
        x = tf.keras.layers.Dense(128, activation='relu')(x)
        logits = tf.keras.layers.Dense(NUM_CLASSES)(x)

    model = tf.keras.Model(inputs, logits)
    return model

# -----------------------
# Training utilities (use GradientTape for backward)
# -----------------------
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
def accuracy_from_logits(logits, labels):
    preds = tf.argmax(logits, axis=-1, output_type=tf.int32)
    # Cast labels to int32 to match preds dtype for tf.equal
    labels = tf.cast(labels, tf.int32)
    return tf.reduce_mean(tf.cast(tf.equal(preds, labels), tf.float32)).numpy()

def train_with_tape(model, x_tr, y_tr, x_val, y_val, epochs=3, batch_size=128):
    ds = tf.data.Dataset.from_tensor_slices((x_tr, y_tr)).shuffle(2000).batch(batch_size)
    optimizer = tf.keras.optimizers.Adam(1e-3)
    epoch_losses = []
    epoch_accs = []
    t_start = time.time()
    for epoch in range(epochs):
        ep_loss = 0.0
        ep_steps = 0
        for (xb, yb) in ds:
            xb_tf = tf.convert_to_tensor(xb)
            yb_tf = tf.convert_to_tensor(yb)
            with tf.GradientTape() as tape:
                logits = model(xb_tf, training=True)
                loss = loss_fn(yb_tf, logits)
            grads = tape.gradient(loss, model.trainable_variables)
            # Some models may include variables created on the fly (gamma/beta) — ensure grads align
            optimizer.apply_gradients(zip(grads, model.trainable_variables))
            ep_loss += loss.numpy()
            ep_steps += 1
        avg_loss = ep_loss / max(1, ep_steps)

        # compute train accuracy on a sample subset (or whole training set if small)
        # to match screenshot style compute accuracy on the training dataset subset
        train_logits = model(tf.convert_to_tensor(x_tr[:2000]), training=False)
        train_acc = accuracy_from_logits(train_logits, tf.convert_to_tensor(y_tr[:2000]))
        epoch_losses.append(avg_loss)
        epoch_accs.append(train_acc * 100.0)  # percentage

        elapsed = time.time() - t_start
        # format lines like screenshot: "Epoch 1/3 | loss: 0.5254 | acc: 81.35%"
        print(f"Epoch {epoch+1}/{epochs} | loss: {avg_loss:.4f} | acc: {train_acc*100:.2f}%")
    total_time = time.time() - t_start
    # final test accuracy
    test_logits = model(tf.convert_to_tensor(x_val), training=False)
    test_acc = accuracy_from_logits(test_logits, tf.convert_to_tensor(y_val)) * 100.0
    print(f"Test Accuracy after training: {test_acc:.2f}%")
    print(f"Time elapsed: {total_time:.1f}s\n")
    return epoch_losses, epoch_accs, test_acc, total_time

# -----------------------
# Run experiments & collect metrics
# -----------------------
EPOCHS = 3
BATCH = 128

all_histories = {}  # store losses and accs for plotting
results_summary = {}

# 1) Baseline
print(">>> Baseline (no normalization)")
baseline = build_baseline()
bl_losses, bl_accs, bl_test_acc, bl_time = train_with_tape(baseline, x_train, y_train, x_test, y_test, epochs=EPOCHS, batch_size=BATCH)
all_histories['Baseline'] = (bl_losses, bl_accs)
results_summary['Baseline'] = (bl_test_acc, bl_time)

# 2) BatchNorm (custom)
print(">>> BatchNorm")
custom_bn = build_custom(norm_type='batch')
bn_losses, bn_accs, bn_test_acc, bn_time = train_with_tape(custom_bn, x_train, y_train, x_test, y_test, epochs=EPOCHS, batch_size=BATCH)
all_histories['BatchNorm'] = (bn_losses, bn_accs)
results_summary['BatchNorm'] = (bn_test_acc, bn_time)

# 3) LayerNorm (custom)
print(">>> LayerNorm")
custom_ln = build_custom(norm_type='layer')
ln_losses, ln_accs, ln_test_acc, ln_time = train_with_tape(custom_ln, x_train, y_train, x_test, y_test, epochs=EPOCHS, batch_size=BATCH)
all_histories['LayerNorm'] = (ln_losses, ln_accs)
results_summary['LayerNorm'] = (ln_test_acc, ln_time)

# 4) WeightNorm (custom WeightNormDense)
print(">>> WeightNorm")
custom_wn = build_custom(norm_type='weight')
wn_losses, wn_accs, wn_test_acc, wn_time = train_with_tape(custom_wn, x_train, y_train, x_test, y_test, epochs=EPOCHS, batch_size=BATCH)
all_histories['WeightNorm'] = (wn_losses, wn_accs)
results_summary['WeightNorm'] = (wn_test_acc, wn_time)

# -----------------------
# Print compact textual summary similar to screenshot
# -----------------------
print("=== Summary Results ===")
for name, (test_acc, tme) in results_summary.items():
    print(f"{name} Test Acc: {test_acc:.2f}%   Time elapsed: {tme:.1f}s")

# -----------------------
# Plot Training Loss and Training Accuracy (like screenshot)
# -----------------------
epochs_range = np.arange(EPOCHS)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
for name, (losses, accs) in all_histories.items():
    plt.plot(epochs_range, losses, label=name)
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.subplot(1,2,2)
for name, (losses, accs) in all_histories.items():
    plt.plot(epochs_range, accs, label=name)
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# -----------------------
# Final short findings print (console)
# -----------------------
print("\nFindings (quick):")
print("- BatchNorm and LayerNorm typically improve convergence vs baseline in this small CNN.")
print("- WeightNorm may help optimization but needs careful integration (and more training).")
print("- LayerNorm is robust for small batch sizes since it does not use batch statistics.")
print("- For rigorous comparisons, run more epochs and multiple seeds; here we show quick 3-epoch runs.")
```
